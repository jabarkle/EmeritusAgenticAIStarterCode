```python
import json
import ollama

# 1. Change your prompt here!
prompt = "How do I reset my password?"

# 2. Get the model's response
response = ollama.chat(model="qwen3:4b", messages=[
    {"role": "user", "content": prompt}
])
assistant_reply = response['message']['content']

# 3. Format as a single training example
record = {
    "messages": [
        {"role": "system", "content": "You are a support agent."},
        {"role": "user", "content": prompt},
        {"role": "assistant", "content": assistant_reply}
    ]
}

# 4. Append to the JSONL file (adds a new line each time it runs)
with open("class_dataset.jsonl", "a") as f:
    f.write(json.dumps(record) + "\n")

print("Successfully saved to JSONL!")
```
